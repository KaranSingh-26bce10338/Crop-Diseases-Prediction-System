```python

"""
Crop Disease Prediction System

***Requires : numpy   (pip install numpy)***

Uses NumPy to compare the symptoms selected by the user
with the symptoms associated with known crop diseases.
"""

import numpy as np


# ------------------------------
# Crop and disease information
# ------------------------------  

SYMPTOM_LIBRARY = {
    "Tomato": {
        "s1": "Small dark brown spots with concentric rings on older leaves",
        "s2": "Yellow halo surrounding leaf spots",
        "s3": "Water-soaked patches that turn brown or black rapidly",
        "s4": "White fuzzy growth on the underside of leaves",
        "s5": "Leaves curling upward and turning yellow",
        "s6": "Stunted plant growth",
        "s7": "Fruit showing dark leathery sunken patches",
    },
    "Wheat": {
        "s1": "Orange/rust-colored powdery pustules on leaves",
        "s2": "Yellow streaks running parallel to leaf veins",
        "s3": "Black powdery masses replacing grains in the ear head",
        "s4": "White powdery coating on leaf surface",
        "s5": "Leaves drying from the tip downward",
        "s6": "Stunted, weak stems that lodge easily",
    },
    "Potato": {
        "s1": "Water-soaked dark green/black patches on leaves",
        "s2": "White mold ring visible on the underside of leaf spots",
        "s3": "Concentric brown rings on leaves",
        "s4": "Foul smell from rotting tubers",
        "s5": "Yellowing and wilting of lower leaves",
        "s6": "Dark sunken lesions on the tuber skin",
    },
    "Rice": {
        "s1": "Diamond/spindle-shaped grey-centered spots on leaves",
        "s2": "Yellow-orange stripes along leaf veins",
        "s3": "Leaves turning white or straw colored and drying",
        "s4": "Small brown oval spots scattered on leaves",
        "s5": "Rotting of the stem base near water level",
        "s6": "Empty or partially filled grains in the panicle",
    },
}


CROP_DISEASE_DB = {
    "Tomato": {
        "Early Blight": {
            "symptoms": {"s1", "s2", "s6"},
            "treatment": (
                "Apply a copper-based or chlorothalonil fungicide every 7-10 days."
            ),
            "prevention": "Rotate crops yearly and avoid overhead watering.",
            "severity": "Medium",
        },
        "Late Blight": {
            "symptoms": {"s3", "s4", "s7"},
            "treatment": (
                "Apply metalaxyl or mancozeb fungicide immediately "
                "and remove infected plants."
            ),
            "prevention": "Ensure good field drainage and spacing for airflow.",
            "severity": "High",
        },
        "Tomato Yellow Leaf Curl": {
            "symptoms": {"s5", "s6"},
            "treatment": (
                "Remove infected plants and control whitefly vectors with neem oil."
            ),
            "prevention": (
                "Use insect-proof nets and whitefly-resistant varieties."
            ),
            "severity": "High",
        },
    },

    "Wheat": {
        "Wheat Rust": {
            "symptoms": {"s1", "s2", "s5"},
            "treatment": (
                "Spray a propiconazole-based fungicide at the first sign of pustules."
            ),
            "prevention": "Plant rust-resistant wheat varieties.",
            "severity": "High",
        },
        "Loose Smut": {
            "symptoms": {"s3", "s6"},
            "treatment": (
                "There is no cure once symptoms are visible. "
                "Remove and destroy infected ear heads."
            ),
            "prevention": (
                "Use hot-water or fungicide seed treatment before sowing."
            ),
            "severity": "Medium",
        },
        "Powdery Mildew": {
            "symptoms": {"s4", "s5"},
            "treatment": "Apply a sulfur-based fungicide spray.",
            "prevention": (
                "Avoid excess nitrogen fertilizer and dense sowing."
            ),
            "severity": "Low",
        },
    },

    "Potato": {
        "Late Blight": {
            "symptoms": {"s1", "s2", "s5"},
            "treatment": (
                "Apply mancozeb or copper oxychloride fungicide immediately."
            ),
            "prevention": "Use certified disease-free seed tubers.",
            "severity": "High",
        },
        "Early Blight": {
            "symptoms": {"s3", "s5"},
            "treatment": (
                "Apply chlorothalonil fungicide and remove infected lower leaves."
            ),
            "prevention": "Practice a 2-3 year crop rotation.",
            "severity": "Medium",
        },
        "Bacterial Soft Rot": {
            "symptoms": {"s4", "s6"},
            "treatment": (
                "Remove and destroy affected tubers and improve storage ventilation."
            ),
            "prevention": (
                "Avoid injuring tubers during harvest and keep storage areas dry."
            ),
            "severity": "Medium",
        },
    },

    "Rice": {
        "Rice Blast": {
            "symptoms": {"s1", "s3"},
            "treatment": (
                "Spray tricyclazole fungicide at the early symptom stage."
            ),
            "prevention": (
                "Avoid excess nitrogen and use blast-resistant varieties."
            ),
            "severity": "High",
        },
        "Bacterial Leaf Blight": {
            "symptoms": {"s2", "s3", "s6"},
            "treatment": (
                "Apply a copper-based bactericide and drain excess field water."
            ),
            "prevention": (
                "Use resistant varieties and clean irrigation water."
            ),
            "severity": "High",
        },
        "Brown Spot": {
            "symptoms": {"s4", "s6"},
            "treatment": (
                "Apply mancozeb fungicide and improve soil potassium levels."
            ),
            "prevention": (
                "Use balanced fertilization and treat seeds before sowing."
            ),
            "severity": "Low",
        },
        "Sheath Blight": {
            "symptoms": {"s5", "s6"},
            "treatment": (
                "Apply hexaconazole fungicide at the base of affected plants."
            ),
            "prevention": (
                "Avoid dense planting to reduce humidity around the stems."
            ),
            "severity": "Medium",
        },
    },
}


def get_supported_crops():
    """Return all crops supported by the system."""
    return tuple(CROP_DISEASE_DB.keys())


def get_symptom_library(crop):
    """Return the symptoms available for a particular crop."""
    return SYMPTOM_LIBRARY.get(crop, {})


def get_diseases_for_crop(crop):
    """Return the known diseases for a particular crop."""
    return CROP_DISEASE_DB.get(crop, {})


# ------------------------- 
# Crop and disease classes
# ------------------------- 

class Plant:
    """Basic information about a plant."""

    def __init__(self, name, plant_type="Crop"):
        self._name = name
        self._plant_type = plant_type

    def get_name(self):
        return self._name

    def describe(self):
        return f"{self._name} is a {self._plant_type}."


class Disease:
    """Stores the information associated with one disease."""

    def __init__(self, name, symptoms, treatment, prevention, severity):
        self._name = name
        self._symptoms = set(symptoms)
        self._treatment = treatment
        self._prevention = prevention
        self._severity = severity

    def get_name(self):
        return self._name

    def get_symptoms(self):
        return self._symptoms

    def get_treatment(self):
        return self._treatment

    def get_prevention(self):
        return self._prevention

    def get_severity(self):
        return self._severity


class Crop(Plant):
    """Represents a crop and the diseases associated with it."""

    def __init__(self, name, symptom_library, disease_data):
        super().__init__(name)
        self._symptom_library = symptom_library
        self._all_symptom_codes = tuple(sorted(symptom_library.keys()))

        self._diseases = []

        for disease_name, disease_info in disease_data.items():
            disease = Disease(
                name=disease_name,
                symptoms=disease_info["symptoms"],
                treatment=disease_info["treatment"],
                prevention=disease_info["prevention"],
                severity=disease_info["severity"],
            )
            self._diseases.append(disease)

    def describe(self):
        return (
            f"{self._name}: tracking "
            f"{len(self._diseases)} known disease(s)."
        )

    def get_symptom_codes(self):
        return self._all_symptom_codes

    def get_diseases(self):
        return self._diseases


# ------------------ 
# Prediction engine
# ------------------ 

class DiseasePredictionEngine:
    """
    Compares the user's symptoms with known disease symptoms.

    Each symptom set is converted into a binary NumPy vector.
    Cosine similarity is then used to calculate how closely
    the selected symptoms match each disease.
    """

    def __init__(self, crop):
        self._crop = crop

    def _vectorize(self, symptoms):
        symptom_codes = self._crop.get_symptom_codes()

        return np.array([
            1 if code in symptoms else 0
            for code in symptom_codes
        ])

    @staticmethod
    def _cosine_similarity(vector_a, vector_b):
        dot_product = np.dot(vector_a, vector_b)

        magnitude_a = np.linalg.norm(vector_a)
        magnitude_b = np.linalg.norm(vector_b)

        if magnitude_a == 0 or magnitude_b == 0:
            return 0.0

        return float(
            dot_product / (magnitude_a * magnitude_b)
        )

    def predict(self, observed_symptoms):
        """Return diseases ranked by symptom similarity."""

        user_vector = self._vectorize(set(observed_symptoms))
        predictions = []

        for disease in self._crop.get_diseases():
            disease_vector = self._vectorize(disease.get_symptoms())

            similarity = self._cosine_similarity(
                user_vector,
                disease_vector
            )

            confidence = round(similarity * 100, 1)
            predictions.append((disease, confidence))

        predictions.sort(
            key=lambda result: result[1],
            reverse=True
        )

        return predictions


# ----------------------------------- 
# Recommendations and session history
# -----------------------------------

prediction_history = []


def severity_warning(severity):
    """Convert the severity level into a simple warning label."""

    labels = {
        "High": "URGENT",
        "Medium": "MODERATE",
        "Low": "MILD",
    }

    return labels.get(severity, "UNKNOWN")


def build_recommendation(crop_name, disease, confidence, mode="detailed"):
    """Create a readable prediction message."""

    warning = severity_warning(disease.get_severity())

    if mode == "brief":
        return (
            f"{warning} | {crop_name}: "
            f"Likely '{disease.get_name()}' "
            f"({confidence}% confidence)"
        )

    return (
        f"\n{'=' * 55}\n"
        f"{warning}  PREDICTION REPORT\n"
        f"{'=' * 55}\n"
        f"Crop             : {crop_name}\n"
        f"Predicted Disease: {disease.get_name()}\n"
        f"Confidence       : {confidence}%\n"
        f"Severity Level   : {disease.get_severity()}\n"
        f"{'-' * 55}\n"
        f"Recommended Treatment:\n"
        f"  -> {disease.get_treatment()}\n"
        f"Prevention for Next Season:\n"
        f"  -> {disease.get_prevention()}\n"
        f"{'=' * 55}\n"
    )


def record_history(crop_name, disease_name, confidence):
    """Save a prediction made during the current session."""

    prediction_history.append(
        (crop_name, disease_name, confidence)
    )


def show_history():
    """Display all predictions made during the current session."""

    if not prediction_history:
        print("\nNo predictions have been made yet this session.\n")
        return

    print(
        f"\n{'#':<3}"
        f"{'Crop':<12}"
        f"{'Disease':<28}"
        f"{'Confidence':<10}"
    )
    print("-" * 53)

    for number, (crop, disease, confidence) in enumerate(
        prediction_history,
        start=1
    ):
        print(
            f"{number:<3}"
            f"{crop:<12}"
            f"{disease:<28}"
            f"{confidence}%"
        )

    print()


# -------------------------------------------------------------------
# User interface
# -------------------------------------------------------------------

def display_welcome():
    print("=" * 55)
    print(" Crop Disease Prediction System")
    print("=" * 55)
    print(
        "This tool compares the symptoms you observe with a "
        "database of known crop diseases."
    )
    print(
        "It uses similarity matching to suggest the most likely "
        "disease and provides basic treatment information.\n"
    )


def select_crop():
    """Ask the user to select a crop."""

    crops = get_supported_crops()

    print("Select a crop:")

    for number, crop in enumerate(crops, start=1):
        print(f"  {number}. {crop}")

    print("  0. Exit Program")

    while True:
        choice = input("Enter choice number: ").strip()

        if not choice.isdigit():
            print("Please enter a valid number.")
            continue

        choice = int(choice)

        if choice == 0:
            return None

        if 1 <= choice <= len(crops):
            return crops[choice - 1]

        print("Choice out of range, try again.")


def collect_symptoms(crop_name):
    """Show available symptoms and collect the user's selections."""

    symptom_library = get_symptom_library(crop_name)
    symptom_codes = sorted(symptom_library.keys())

    print(
        f"\nObserved symptoms for {crop_name} "
        "-- select all that apply."
    )

    for code in symptom_codes:
        print(f"  [{code}] {symptom_library[code]}")

    print(
        "\nEnter symptom codes separated by commas "
        "(example: s1,s3,s5):"
    )

    user_input = input("> ").strip().lower()

    entered_codes = {
        code.strip()
        for code in user_input.split(",")
        if code.strip()
    }

    valid_codes = entered_codes.intersection(symptom_codes)
    invalid_codes = entered_codes - valid_codes

    if invalid_codes:
        print(
            "Ignoring unrecognised codes: "
            + ", ".join(sorted(invalid_codes))
        )

    return valid_codes


def display_top_results(crop_name, predictions):
    """Display the best prediction and one alternative."""

    if not predictions or predictions[0][1] == 0:
        print(
            "\nNo close match found. "
            "Try selecting more or different symptoms.\n"
        )
        return

    best_disease, best_confidence = predictions[0]

    print(
        build_recommendation(
            crop_name,
            best_disease,
            best_confidence
        )
    )

    record_history(
        crop_name,
        best_disease.get_name(),
        best_confidence
    )

    if len(predictions) > 1 and predictions[1][1] > 0:
        alternative, confidence = predictions[1]

        print("Other possible match:")
        print(
            " ",
            build_recommendation(
                crop_name,
                alternative,
                confidence,
                mode="brief"
            )
        )
        print()


def ask_continue():
    """Ask whether the user wants to make another prediction."""

    answer = input(
        "Check another crop? (y/n): "
    ).strip().lower()

    return answer == "y"


# -------------------------------------------------------------------
# Main program
# -------------------------------------------------------------------

def run_prediction_for_crop(crop_name):
    """Run the prediction process for the selected crop."""

    symptom_library = get_symptom_library(crop_name)
    disease_data = get_diseases_for_crop(crop_name)

    crop = Crop(
        crop_name,
        symptom_library,
        disease_data
    )
    print(f"\n{crop.describe()}")
    prediction_engine = DiseasePredictionEngine(crop)
    observed_symptoms = collect_symptoms(crop_name)
    if not observed_symptoms:
        print("No symptoms selected. Skipping prediction.\n")
        return
    predictions = prediction_engine.predict(observed_symptoms)
    display_top_results(
        crop_name,
        predictions
    )
def main():
    display_welcome()
    while True:
        crop_name = select_crop()
        if crop_name is None:
            print("\nSession summary:")
            show_history()
            print(
                "Thank you for using the "
                "Crop Disease Prediction System."
            )
            break
        run_prediction_for_crop(crop_name)
        if not ask_continue():
            print("\nSession summary:")
            show_history()
            print(
                "Thank you for using the "
                "Crop Disease Prediction System."
            )
            break
if __name__ == "__main__":
    main()
```
